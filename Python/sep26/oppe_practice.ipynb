## Count Positive Integers
Given the following list:

numbers = [1, -2, 3, 0, None]

Write a program that counts how many values are positive integers.

Ignore None values.
Ignore 0.
Use continue to skip values that should not be counted.
Print the final count.

```python
n=[1,-2,3,0,None,-4,5]
count=0
for val in n:
  if val is None or val <=0:
    continue
  count+=1
print(count)
```

## Count Words with More Than 4 Letters

Write a program that takes a line of text as input.

The words are separated by spaces.

Count how many words have more than 4 characters. Print the final count.

Input Format

A single line containing words separated by spaces.

Output Format

Print the number of words with more than 4 characters.

**Example**

Input

I love learning Python programming

Output

3

```python
sen = input()
lis = sen.split()
count = 0
for elem in lis :
  if len(elem) > 4 :
    count += 1
print(count)
```

## Problem Statement

Find the Last Word That Starts with an Uppercase Letter

Write a program that takes a line of text as input. The words in the input are separated by spaces.

Find and print the last word whose first character is an uppercase letter.

Input Format

A single line containing words separated by spaces.

Output Format

Print the last word that starts with an uppercase letter.

Example

Input

This is a Test sentence

Output

Test

Assumption: At least one word in the input starts with an uppercase letter.

```python
text=input()
result=""
text_l=text.split()
for word in text_l:
  if word[0].isupper():
    result=word
print(result)
```
