### Tuple

```python
# all the operations performed on lists can be performed on Tuples expect those which involves modifcation. The ones can be performed is count, index, they just iterate over the data structure and gives the answer, we can also perform slicing,
# tupple is immutable
# we can perform slicing
# if tupple contains only one element it is considered as integer
```

```python
t=1,2 # even without () it is tupple the important thing is comma
print(type(t))
```

```python
t=12
print(type(t))
```

```python
x,y=t # unpacking a tupple and packing a tupple is t=(1,2)
print(x,y)
```

```python
l=[(1,92), (2,75), (3,100)]
print(type(l))
for i in l:
  print(i) # here i refers to every element of the list, however the list has tupple as its elements hence i refers to the tupples
```

```python
l=[(1,92), (2,75), (3,100)]
print(type(l))
for i,j in l:
  print(i)
```

```python
l=[(1,92), (2,75), (3,100)]
print(type(l))
for i,j in l:
  print(i,j)
```

```python
l=[(1,92), (2,75), (3,100)]
print(type(l))
for i,j in l:
  print(type(i),type(j))
```

#### Tupple unpacking

```python
# tupple unpacking has to be done correctly for all the elements
l=[(1,92,"A+"), (2,75,"B"), (3,100,"A++")]
print(type(l))
for i,j,z in l:
  print(i,j,z)
```
