## Locall and Global Variables

Global variables are acccessable everywhere

```python
x=10
def change():
  x=20
  print(x)
```

```python
print(x)
change()
```

```python
x=10
def change():
  global x
  x=20
  print(x)
```

```python
print(x)
change()
```

```python
str1="ProgrammingACAT"
str_res=""
for char in str1:
  if char.lower() in "aeiou":
    str_res+=char
print(str_res)
```

```python
str1="ProgrammingACAT"
str_res=""
for char in str1:
  if char.lower() in "aeiou":
    str_res+=char
    print(str_res)
```

```python
str1="ProgrammingACAT"
str_res=""
for char in str1:
  if char.lower() in "aeiou" and char not in "ABCDEFGHIJKLMNOPQRSTUVWXYZ":
    str_res+=char
print(str_res)
```
