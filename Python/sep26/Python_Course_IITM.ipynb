Question 1: Find students who scored above 80. You are given a dictionary containing student names as keys and their marks as vakues.
Create a new list containing the names of students who scored more than 80,
Input students = {"Aman":85, "Rahul":72, "Riya":92, "Karan":65, "Priya": 88}

Expectzed output:['Aman', 'Riya', 'Priya']

```python
st_marks = {"Aman":85, "Rahul":72, "Riya":92, "Karan":65, "Priya":88}
print(st_marks)
l=[]
for key in st_marks:
  if st_marks[key]>80:
    l.append(key)
print(l)
```
