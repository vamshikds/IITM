## key

to sort the list based on the length of the strings

```python
l4=["Apple", "Kiwi", "Banana"]
print(l4[::-1])
```

```python
l4=["Apple", "Kiwi", "Banana"] # we can use iterations and can be little lengthy approach
result= sorted(l4,key=len,reverse=True)
print(result)
print(l4)
```

```python
l5=[1,2,-3,4,-5,-100]
result=sorted(l5,reverse=True)
print(result)
```

```python
def my_fun(word):
  count = 0
  for i in word :
    count += 1
    if i == 'a':
      return count
  return count
l = ['apple','kiwi','banana']
a = sorted(l,key=my_fun)
print(a)
```

## Lambda Function

This function will give error is the index value is missing in the mentioned tsrings or even if it doesnt match the case

```python
l4=["Apple", "Kiwia", "Banana"]
result= sorted(l4,key=lambda x:x.index("a"))
print(result)
print(l4)
```

```python
l4=["apple", "Kiwia", "Banana"]
result= sorted(l4,key=lambda x:x.index("a"))
print(result)
print(l4)
```

```python
l4=["avacado", "apple", "gghsja", "Kiwia", "Banana"]
result= sorted(l4,key=lambda x:x.index("a"))
print(result)
print(l4)
```

```python
l4=["avacado", "apple", "gghsja", "Kiwia", "Banana"]
result= sorted(l4,key=lambda x:(x.index("a"),x)) #first the x indexes the first a in each string so its 0,0,5,4,2
print(result)
print(l4)
```

```python
l4=["avacado", "apple", "gghsja", "Kiwia", "Banana"]
result= sorted(l4,key=lambda x:(x.index("a"),x), reverse=True)
print(result)
print(l4)
```

```python
l6=["Pigeion", "Peacock", "Parrot"]
result= sorted(l6,key=lambda x:(x.index("a"),x)) # after : we need to write something which the function returns
print(result)
print(l6)
```
